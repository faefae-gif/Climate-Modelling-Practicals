```python
import xarray as xr
import numpy as np
import pandas as pd
from pathlib import Path

# Mount Makulu coordinates
STATION_LAT = -15.55
STATION_LON = 28.25

OUTPUT_DIR = Path("mount_makulu_extracted")
OUTPUT_DIR.mkdir(exist_ok=True)

# Store information about each extracted file
extraction_log = []

for i, file in enumerate(nc_files, start=1):

    print(f"[{i}/{len(nc_files)}] Processing: {file.name}")

    ds = xr.open_dataset(file)

    # ---------------------------------------------------------
    # Identify climate variable
    # ---------------------------------------------------------
    if file.name.startswith("tasmax"):
        variable = "tasmax"
    elif file.name.startswith("tasmin"):
        variable = "tasmin"
    elif file.name.startswith("pr_"):
        variable = "pr"
    else:
        ds.close()
        print("   Skipped: unknown variable")
        continue

    # ---------------------------------------------------------
    # Find nearest geographic grid cell to Mount Makulu
    # ---------------------------------------------------------
    lat = ds["lat"].values
    lon = ds["lon"].values

    # Distance in lat/lon space
    distance = np.sqrt(
        (lat - STATION_LAT)**2 +
        (lon - STATION_LON)**2
    )

    iy, ix = np.unravel_index(
        np.nanargmin(distance),
        distance.shape
    )

    nearest_lat = float(lat[iy, ix])
    nearest_lon = float(lon[iy, ix])

    # ---------------------------------------------------------
    # Extract time series
    # ---------------------------------------------------------
    da = ds[variable].isel(
        rlat=iy,
        rlon=ix
    )

    values = da.values.astype(float)

    # ---------------------------------------------------------
    # Unit conversion
    # ---------------------------------------------------------
    units = da.attrs.get("units", "")

    # Temperature: Kelvin -> Celsius
    if variable in ["tasmax", "tasmin"]:
        if units.lower() in ["k", "kelvin"]:
            values = values - 273.15

    # Precipitation:
    # kg m-2 s-1 -> mm/day
    elif variable == "pr":
        if "s-1" in units or "s**-1" in units:
            values = values * 86400.0

    # ---------------------------------------------------------
    # Convert dates
    # ---------------------------------------------------------
    dates = pd.to_datetime([
    f"{t.year:04d}-{t.month:02d}-{t.day:02d}"
    for t in ds.indexes["time"]
])

    # ---------------------------------------------------------
    # Create dataframe
    # ---------------------------------------------------------
    df = pd.DataFrame({
        "date": dates,
        variable: values
    })

    # Save one CSV per NetCDF file
    output_file = OUTPUT_DIR / f"{file.stem}.csv"
    df.to_csv(output_file, index=False)

    # ---------------------------------------------------------
    # Record extraction information
    # ---------------------------------------------------------
    extraction_log.append({
        "file": file.name,
        "variable": variable,
        "nearest_lat": nearest_lat,
        "nearest_lon": nearest_lon,
        "rlat_index": iy,
        "rlon_index": ix,
        "units_original": units,
        "n_days": len(df),
        "missing": int(df[variable].isna().sum())
    })

    ds.close()

print("\n✅ ALL FILES PROCESSED")
```

## DEFINING THE EQM FUNCTIONS

```python
import numpy as np

# ============================================================
# TEMPERATURE EQM
# ============================================================

def eqm_temperature(model_cal, obs_cal, values):

    model_cal = np.asarray(model_cal, dtype=float)
    obs_cal = np.asarray(obs_cal, dtype=float)
    values = np.asarray(values, dtype=float)

    # Remove missing calibration values
    model_cal = model_cal[np.isfinite(model_cal)]
    obs_cal = obs_cal[np.isfinite(obs_cal)]

    model_sorted = np.sort(model_cal)
    obs_sorted = np.sort(obs_cal)

    model_q = np.linspace(0, 1, len(model_sorted))
    obs_q = np.linspace(0, 1, len(obs_sorted))

    # Find percentile of each model value
    percentiles = np.interp(
        values,
        model_sorted,
        model_q,
        left=0,
        right=1
    )

    # Map percentile onto observations
    corrected = np.interp(
        percentiles,
        obs_q,
        obs_sorted
    )

    return corrected


# ============================================================
# PRECIPITATION EQM
# ============================================================

def eqm_precipitation(
    model_cal,
    obs_cal,
    values,
    wet_threshold=0.1
):

    model_cal = np.asarray(model_cal, dtype=float)
    obs_cal = np.asarray(obs_cal, dtype=float)
    values = np.asarray(values, dtype=float)

    model_cal = model_cal[np.isfinite(model_cal)]
    obs_cal = obs_cal[np.isfinite(obs_cal)]

    corrected = np.zeros(len(values), dtype=float)

    # Preserve missing values
    missing = ~np.isfinite(values)
    corrected[missing] = np.nan

    # Observed wet days
    obs_wet = obs_cal[
        obs_cal >= wet_threshold
    ]

    # If there are no observed wet days in a month
    if len(obs_wet) == 0:
        corrected[~missing] = 0.0
        return corrected

    # Observed wet-day frequency
    obs_wet_fraction = (
        len(obs_wet) / len(obs_cal)
    )

    # Model threshold giving equivalent wet-day frequency
    model_threshold = np.quantile(
        model_cal,
        1 - obs_wet_fraction
    )

    wet_values = (
        np.isfinite(values) &
        (values > model_threshold)
    )

    if wet_values.sum() == 0:
        return corrected

    model_wet = model_cal[
        model_cal > model_threshold
    ]

    if len(model_wet) == 0:
        return corrected

    model_sorted = np.sort(model_wet)
    obs_sorted = np.sort(obs_wet)

    model_q = np.linspace(
        0, 1, len(model_sorted)
    )

    obs_q = np.linspace(
        0, 1, len(obs_sorted)
    )

    percentiles = np.interp(
        values[wet_values],
        model_sorted,
        model_q,
        left=0,
        right=1
    )

    corrected[wet_values] = np.interp(
        percentiles,
        obs_q,
        obs_sorted
    )

    return corrected


print("✅ EQM functions ready")
```

```python
corrected_calibration = {}

# Add month to observations
obs_eqm = obs.copy()
obs_eqm["month"] = obs_eqm["date"].dt.month

for model_name in calibration_data:

    corrected_calibration[model_name] = {}

    for scenario in ["rcp45", "rcp85"]:

        print(f"Correcting {model_name} - {scenario}")

        df = calibration_data[model_name][scenario].copy()
        df["month"] = df["date"].dt.month

        # Columns for corrected values
        df["tasmax_bc"] = np.nan
        df["tasmin_bc"] = np.nan
        df["pr_bc"] = np.nan

        # ----------------------------------------------------
        # Correct each calendar month separately
        # ----------------------------------------------------
        for month in range(1, 13):

            model_mask = df["month"] == month
            obs_month = obs_eqm[obs_eqm["month"] == month]

            # Tmax
            df.loc[model_mask, "tasmax_bc"] = eqm_temperature(
                df.loc[model_mask, "tasmax"].values,
                obs_month["tmax"].dropna().values,
                df.loc[model_mask, "tasmax"].values
            )

            # Tmin
            df.loc[model_mask, "tasmin_bc"] = eqm_temperature(
                df.loc[model_mask, "tasmin"].values,
                obs_month["tmin"].dropna().values,
                df.loc[model_mask, "tasmin"].values
            )

            # Precipitation
            df.loc[model_mask, "pr_bc"] = eqm_precipitation(
                df.loc[model_mask, "pr"].values,
                obs_month["precip"].dropna().values,
                df.loc[model_mask, "pr"].values,
                wet_threshold=0.1
            )

        corrected_calibration[model_name][scenario] = df

print("\n✅ Monthly EQM completed for all 6 models!")
```

## Add the future-temperature EQM function

```python
def eqm_temperature_future(model_cal, obs_cal, future_values):

    model_cal = np.asarray(model_cal, dtype=float)
    obs_cal = np.asarray(obs_cal, dtype=float)
    future_values = np.asarray(future_values, dtype=float)

    model_cal = model_cal[np.isfinite(model_cal)]
    obs_cal = obs_cal[np.isfinite(obs_cal)]

    model_sorted = np.sort(model_cal)
    obs_sorted = np.sort(obs_cal)

    model_q = np.linspace(0, 1, len(model_sorted))
    obs_q = np.linspace(0, 1, len(obs_sorted))

    corrected = np.full(len(future_values), np.nan)

    valid = np.isfinite(future_values)

    # Standard quantile mapping
    percentiles = np.interp(
        future_values[valid],
        model_sorted,
        model_q,
        left=0,
        right=1
    )

    corrected[valid] = np.interp(
        percentiles,
        obs_q,
        obs_sorted
    )

    # Additive extrapolation above calibration maximum
    above = valid & (future_values > model_sorted[-1])

    corrected[above] = (
        obs_sorted[-1]
        + (future_values[above] - model_sorted[-1])
    )

    # Additive extrapolation below calibration minimum
    below = valid & (future_values < model_sorted[0])

    corrected[below] = (
        obs_sorted[0]
        + (future_values[below] - model_sorted[0])
    )

    return corrected


print("✅ Future temperature EQM ready")
```

## Create the calibration climatology

```python
# ============================================================
# CALIBRATION CLIMATOLOGY
# Raw vs Bias-Corrected vs Observed
# 1981–2010
# ============================================================

calibration_rows = []

for model_name in corrected_calibration:

    # Use RCP4.5 calibration branch
    df = corrected_calibration[model_name]["rcp45"].copy()

    df["month"] = df["date"].dt.month

    monthly = (
        df.groupby("month")
          .agg(
              tasmax_raw=("tasmax", "mean"),
              tasmax_bc=("tasmax_bc", "mean"),

              tasmin_raw=("tasmin", "mean"),
              tasmin_bc=("tasmin_bc", "mean"),

              pr_raw=("pr", "mean"),
              pr_bc=("pr_bc", "mean")
          )
    )

    for month in range(1, 13):

        calibration_rows.append({
            "Model": model_name,
            "Month": month,

            "Tmax_Raw": monthly.loc[month, "tasmax_raw"],
            "Tmax_BC": monthly.loc[month, "tasmax_bc"],

            "Tmin_Raw": monthly.loc[month, "tasmin_raw"],
            "Tmin_BC": monthly.loc[month, "tasmin_bc"],

            "Precip_Raw": monthly.loc[month, "pr_raw"],
            "Precip_BC": monthly.loc[month, "pr_bc"]
        })


calibration_models = pd.DataFrame(calibration_rows)


# ============================================================
# MULTI-MODEL MEAN
# ============================================================

calibration_ensemble = (
    calibration_models
    .groupby("Month")
    .agg(
        Tmax_Raw=("Tmax_Raw", "mean"),
        Tmax_BC=("Tmax_BC", "mean"),

        Tmin_Raw=("Tmin_Raw", "mean"),
        Tmin_BC=("Tmin_BC", "mean"),

        Precip_Raw=("Precip_Raw", "mean"),
        Precip_BC=("Precip_BC", "mean")
    )
    .reset_index()
)


# ============================================================
# ADD OBSERVED CLIMATOLOGY
# ============================================================

calibration_ensemble["Tmax_Obs"] = (
    calibration_ensemble["Month"]
    .map(obs_monthly["tmax"])
)

calibration_ensemble["Tmin_Obs"] = (
    calibration_ensemble["Month"]
    .map(obs_monthly["tmin"])
)

calibration_ensemble["Precip_Obs"] = (
    calibration_ensemble["Month"]
    .map(obs_monthly["precip"])
)


print("✅ Calibration climatology created")

print(
    calibration_ensemble
    .round(2)
    .to_string(index=False)
)
```

```python
# ============================================================
# ANNUAL TEMPERATURE SERIES: 2041–2070
# ============================================================

annual_rows = []

for model_name in future_corrected:

    for scenario in ["rcp45", "rcp85"]:

        df = future_corrected[model_name][scenario].copy()

        df["Year"] = df["date"].dt.year

        annual = (
            df.groupby("Year")
              .agg(
                  Tmax=("tasmax_bc", "mean"),
                  Tmin=("tasmin_bc", "mean")
              )
              .reset_index()
        )

        for _, row in annual.iterrows():

            annual_rows.append({
                "Model": model_name,
                "Scenario": scenario.upper(),
                "Year": int(row["Year"]),
                "Tmax": row["Tmax"],
                "Tmin": row["Tmin"]
            })

annual_models = pd.DataFrame(annual_rows)

# Six-model ensemble mean for each year
annual_ensemble = (
    annual_models
    .groupby(["Scenario", "Year"])
    .agg(
        Tmax=("Tmax", "mean"),
        Tmin=("Tmin", "mean")
    )
    .reset_index()
)

print("✅ Annual temperature series created")

print(
    annual_ensemble
    .head(10)
    .round(2)
    .to_string(index=False)
)
```

```python
from scipy.stats import linregress

trend_rows = []

for scenario in ["RCP45", "RCP85"]:

    df = annual_ensemble[
        annual_ensemble["Scenario"] == scenario
    ]

    # Tmax trend
    tmax_trend = linregress(
        df["Year"],
        df["Tmax"]
    )

    # Tmin trend
    tmin_trend = linregress(
        df["Year"],
        df["Tmin"]
    )

    trend_rows.append({
        "Scenario": scenario,
        "Tmax_Trend_C_per_decade": tmax_trend.slope * 10,
        "Tmax_p_value": tmax_trend.pvalue,
        "Tmin_Trend_C_per_decade": tmin_trend.slope * 10,
        "Tmin_p_value": tmin_trend.pvalue
    })

temperature_trends = pd.DataFrame(trend_rows)

print(
    temperature_trends
    .round(4)
    .to_string(index=False)
)

print("\n✅ Temperature trends calculated")
```
